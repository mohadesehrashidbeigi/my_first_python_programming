```python
vazn=float(input("وزن خود را وارد کنید:"))
ghad=float(input("قد خود را وارد کنید"))
ghad= ghad/100
bmi=vazn/(ghad**2)
print("شما برابر است باBMI", bmi)
if bmi <18.5:
    print("شما کمبود وزن دارید")
elif bmi <=24.9:
    print("وزن شما ایده آل است")
else:
    print("شما دارای اضافه وزن هستید", round(bmi, 2)) 

```
